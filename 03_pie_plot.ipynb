```python
# plt.pie(values, label = "label_list", color = "color_name", autopct = "%1.1f%%")
import matplotlib.pyplot as plt

regions = ["East", "West", "North", "South"]
revenue = [3000, 2000, 1500, 1000]

plt.pie(revenue, labels=regions, colors=["gold", "skyblue", "lightgreen", "coral"], autopct = "%1.1f%%", explode = [0.1, 0, 0, 0], shadow= True, startangle=90, radius=2)
plt.title("Sales in Different Region")
plt.legend(title = "2026 Sales")
plt.show()

```
