### plt.subplot(nrows, ncols, index)
##### nrows - numbers of rows
##### ncols - nubmers of columns
##### index - position of subplot (start from 1) 

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [25, 34, 26, 40, 54]

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("Line Chart")

plt.subplot(1, 2, 2)
plt.bar(x, y)
plt.title("Bar Chart")

plt.tight_layout()

plt.show()
```

### Industry approach to used subplot

#### Syntax
##### figure, axis = plt.subplots(nrows, ncols, figsize = (width, height))


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [25, 34, 26, 40, 54]

fig, ax = plt.subplots(1, 2, figsize = (10, 5))

ax[0].plot(x, y)
ax[0].set_title("Line Chart")

ax[1].bar(x, y)
ax[1].set_title("Bar Chart")

plt.tight_layout()
plt.suptitle("Comparison of Line and Bar Chart")

plt.show()
```
