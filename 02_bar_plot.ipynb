```python
# plt.bar(x-axis_value, height of bar (generally it is y-axis value), color = "color_name", label = "label_name")

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
sales = [400, 350, 300, 450]


plt.bar(fruits, sales)
plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```

```python
departments = ["IT", "HR", "Finance", "Sales"]
employees = [50, 20, 30, 40]

plt.bar(departments, employees, color = "pink")
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Employees")
# plt.grid()
plt.show()
```

# Horizontal Bar PLot

```python
plt.barh(departments, employees)
```

# Multiple Bar Plot (Grouped Bar Chart)

```python
subjects = ["Math", "Science", "English"]

student1 = [80, 70, 90]
student2 = [75, 85, 88]

x = np.arange(len(subjects))

plt.bar(x - 0.2, student1, width=0.4, label="Student 1")
plt.bar(x + 0.2, student2, width=0.4, label="Student 2")

plt.xticks(x, subjects)
plt.legend()

plt.show()
```

```python
boys = [30, 25, 35]
girls = [20, 15, 25]

classes = ["A", "B", "C"]

plt.bar(classes, boys, label="Boys")
plt.bar(classes, girls, label="Girls")

plt.legend()
plt.show()
```
