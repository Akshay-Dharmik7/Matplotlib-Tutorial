```python
# plt.scatter(x-axis_value, y-axis_value, color = "color_name", marker = "marker symbol", label = "label_name")
import matplotlib.pyplot as plt

Hours = [1, 2, 3, 4, 5, 6, 7, 8]
Scores = [55, 60, 65, 70, 75, 80, 85, 90]

plt.scatter(Hours, Scores, color = "orange", s = 200, marker= "o", label = "Score Data", alpha = 0.5, cmap= "nipy_spectral")

plt.title("Relationship Between Hours and exam Scores")
plt.xlabel = "Hours Studied"
plt.ylabel = "Exam Scores"
plt.legend()
plt.grid(True)
plt.colorbar()

plt.show()
```
