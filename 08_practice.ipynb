```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Univariate Analysis -- Single Variable

```python
data = pd.read_csv('tancent.csv')
data.head()
```

```python
data["Genre"]
```

### Categorical

```python
data['Genre'].unique()
```

```python
cat_counts = data["Genre"].value_counts()
cat_counts
```

I can find the Top 5 Genre

```python
x_bar = cat_counts.index
```

```python
x_bar
```

```python
y_bar = cat_counts.values
```

```python
y_bar
```

### Bar Chart - Matplotlib

```python
plt.bar(x_bar, y_bar)
```

```python
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar)
```

```python
plt.figure(figsize=(12,8))
plt.xticks(rotation = 90)
plt.bar(x_bar, y_bar)
plt.xlabel('Genre', fontsize = 12)
plt.ylabel('Count', fontsize = 12)
plt.title("Top Genre Analysis", fontsize = 18)
plt.show()
```

### Countplot - Seaborn

```python

plt.figure(figsize=(12,8))
plt.xticks(rotation = 90)
sns.countplot(x='Genre', data=data, order=data["Genre"].value_counts().index, color = 'orange')
plt.xlabel('Genre', fontsize = 12)
plt.ylabel('Count', fontsize = 12)
plt.title("Top Genre Analysis", fontsize = 18)
```

### Percentage of sales across the regions

```python
sales_data = data[['NA_Sales','EU_Sales','JP_Sales','Other_Sales']]
```

```python
sales_data
```

```python
region_sales = sales_data.T.sum(axis='columns')
```

```python
region_sales
```

```python
plt.pie(region_sales, labels=region_sales.index, startangle=90, explode=(0.2, 0, 0, 0), autopct='%1.1f%%')
plt.title('Total Sales across various Reions')
plt.show()
```
