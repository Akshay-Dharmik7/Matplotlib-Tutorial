#### Save chart in local 

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [25, 34, 26, 40, 54]
plt.plot(x, y, color = "orange", marker = "o", label = "Plot Label")
plt.title("Line Chart")
plt.xlabel("X-Label")
plt.ylabel("Y-Label")
plt.legend()

# saving chart in local storage
plt.savefig("line_chart.png", dpi = 300, bbox_inches = "tight", transparent = True, facecolor = "skyblue")
plt.show()
```
