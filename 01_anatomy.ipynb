```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 120, 150, 130]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(months, sales, marker='o', label='Sales')

ax.set_title("Monthly Sales Report")
ax.set_xlabel("Months")
ax.set_ylabel("Sales")
ax.grid(True)
ax.legend()

plt.show()
```

```python
# plt.plot() function
# plt.plot(x-axis_value, y-axis_value, color = "color name", linestyle = "line_style", linewidth = value, marker = "marker symbol", label = "label name")

import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]
sales = [1000, 1500, 1200, 2200, 1900]

plt.plot(months, sales, color = "purple", linestyle = "--", linewidth = 2, marker = "o", label = "2026 Sales by Month")
plt.title("Monthly Sales Data Report")
plt.xlabel = "Months"
plt.ylabel = "Sales per Month"
plt.legend(loc = "upper left", fontsize = 10) #loc = "upper left or lower right"
plt.grid(color = "grey", linestyle = ":", linewidth = 1)
plt.xlim(1, 5)
plt.ylim(500, 2500)
plt.xticks([1, 2,3, 4, 5], ["January", "February", "March", "April", "May"])
# plt.yticks()
plt.show()
```
